```python
# Проверка NULL

def check_nulls(table_name):

    # Получаем список столбцов таблицы
    columns = pd.read_sql(
        f"PRAGMA table_info({table_name});",
        engine
    )["name"].tolist()


    # Получаем количество строк
    rows_count = pd.read_sql(
        f"SELECT COUNT(*) AS cnt FROM {table_name};",
        engine
    ).loc[0, "cnt"]


    # Формируем один SQL-запрос
    null_checks = []

    for column in columns:

        null_checks.append(
            f"""
            SUM(
                CASE
                    WHEN {column} IS NULL THEN 1
                    ELSE 0
                END
            ) AS {column}
            """
        )


    sql = f"""
    SELECT
        {",".join(null_checks)}
    FROM {table_name};
    """


    # Выполняем один проход по таблице
    result = pd.read_sql(sql, engine)


    # Переводим широкий формат в удобный
    result = result.T.reset_index()

    result.columns = [
        "column",
        "null_count"
    ]


    result["table"] = table_name

    result["null_percent"] = (
        result["null_count"]
        /
        rows_count
        *
        100
    ).round(2)


    return result[
        [
            "table",
            "column",
            "null_count",
            "null_percent"
        ]
    ]
```

```python
nulls_check = pd.concat(
    [
        check_nulls(table)
        for table in table_list
    ],
    ignore_index=True
)

nulls_check
```

```python
# Дубликаты

def check_duplicates(table_name, column):

    # Ищем только значения ключа,
    # которые встречаются больше одного раза.
    # Это позволяет не считать все DISTINCT значения.

    sql = f"""
    SELECT
        COUNT(*) AS duplicate_rows
    FROM
    (
        SELECT
            {column}
        FROM {table_name}
        GROUP BY {column}
        HAVING COUNT(*) > 1
    );
    """


    result = pd.read_sql(sql, engine)


    # Получаем общее количество строк
    sql_total = f"""
    SELECT COUNT(*) AS total_rows
    FROM {table_name};
    """

    total_rows = pd.read_sql(
        sql_total,
        engine
    ).loc[0, "total_rows"]


    result["table"] = table_name
    result["column"] = column

    result["total_rows"] = total_rows

    result["duplicate_percent"] = (
        result["duplicate_rows"]
        /
        total_rows
        *
        100
    ).round(2)


    return result[
        [
            "table",
            "column",
            "total_rows",
            "duplicate_rows",
            "duplicate_percent"
        ]
    ]
```

#  1. Анализ продаж и ассортимента продуктов питания

```python
product_sales_analysis = pd.read_sql(
    """
    WITH product_sales AS (

        -- продажи по каждому товару

        SELECT

            product_id,

            COUNT(DISTINCT order_id) AS orders_count,

            COUNT(product_id) AS units_sold

        FROM order_products__prior

        GROUP BY product_id

    )


    SELECT

        p.product_id,

        p.product_name,

        d.department,


        -- продажи

        COALESCE(ps.orders_count,0) AS orders_count,

        COALESCE(ps.units_sold,0) AS units_sold,


        -- финансовые показатели

        COALESCE(
            ps.units_sold * pf.price,
            0
        ) AS revenue,


        COALESCE(
            ps.units_sold * pf.cogs,
            0
        ) AS cogs,


        COALESCE(
            ps.units_sold * (pf.price - pf.cogs),
            0
        ) AS gross_profit


    FROM products p


    LEFT JOIN product_sales ps

        ON p.product_id = ps.product_id


    LEFT JOIN departments d

        ON p.department_id = d.department_id


    LEFT JOIN product_financials pf

        ON p.product_id = pf.product_id

    """,
    engine
)


product_sales_analysis.head()
```

```python
# Маржинальность товара

product_sales_analysis["margin"] = np.where(
    product_sales_analysis["revenue"] > 0,
    (
        product_sales_analysis["gross_profit"]
        /
        product_sales_analysis["revenue"]
        * 100
    ).round(2),
    0
)
```

```python
# Статус товара

product_sales_analysis["sales_status"] = np.where(
    product_sales_analysis["units_sold"] > 0,
    "Продается",
    "Нет продаж"
)
```

## 1.1. Структура и эффективность ассортимента

```python
# Общее количество товаров

print(
    "Количество товаров:",
    product_sales_analysis["product_id"].nunique()
)
```

```python
# Общее количество категорий

print(
    "Количество категорий  товаров:",
    product_sales_analysis["department"].nunique()
)

print(
    "Категории товаров:",
    product_sales_analysis["department"].unique()
)
```

```python
# Анализ эффективности категорий

category_efficiency = (
    product_sales_analysis
    .groupby("department")
    .agg(
        products=("product_id","count"),
        units_sold=("units_sold","sum"),
        revenue=("revenue","sum"),
        profit=("gross_profit","sum")
    )
)


# Доля категории в ассортименте

category_efficiency["assortment_share"] = (
    category_efficiency["products"]
    /
    category_efficiency["products"].sum()
    * 100
).round(2)


# Доля выручки

category_efficiency["revenue_share"] = (
    category_efficiency["revenue"]
    /
    category_efficiency["revenue"].sum()
    * 100
).round(2)


# Доля прибыли

category_efficiency["profit_share"] = (
    category_efficiency["profit"]
    /
    category_efficiency["profit"].sum()
    * 100
).round(2)

# Эффективность ассортимента:
# сколько доли прибыли приносит 1% ассортимента

category_efficiency["assortment_efficiency"] = (
    category_efficiency["profit_share"]
    /
    category_efficiency["assortment_share"]
).round(2)


# Маржинальность

category_efficiency["margin"] = (
    category_efficiency["profit"]
    /
    category_efficiency["revenue"]
    * 100
).round(2)


# Перевод в миллионы

category_efficiency["revenue_mln"] = (
    category_efficiency["revenue"] / 1_000_000
).round(2)


category_efficiency["profit_mln"] = (
    category_efficiency["profit"] / 1_000_000
).round(2)
```

```python
# копия исходной таблицы
abc_profit = product_sales_analysis.copy()


# сортируем товары по валовой прибыли
abc_profit = (
    abc_profit
    .sort_values(
        "gross_profit",
        ascending=False
    )
)


# доля товара в общей прибыли

abc_profit["profit_share"] = (
    abc_profit["gross_profit"]
    /
    abc_profit["gross_profit"].sum()
)


# накопительная доля прибыли

abc_profit["cumulative_profit_share"] = (
    abc_profit["profit_share"]
    .cumsum()
)


# присваиваем ABC-сегмент

abc_profit["ABC_segment"] = np.where(
    abc_profit["cumulative_profit_share"] <= 0.8,
    "A",
    np.where(
        abc_profit["cumulative_profit_share"] <= 0.95,
        "B",
        "C"
    )
)


# товары без продаж выделяем отдельно

abc_profit.loc[
    abc_profit["units_sold"] == 0,
    "ABC_segment"
] = "Нет продаж"


# Проверяем распределение товаров

abc_profit["ABC_segment"].value_counts()
```

#  2. Юнит-экономика заказа

```python
# Объединяем данные о заказах, товарах,
# финансовых показателях товаров и расходах по заказам

order_unit_economics = pd.read_sql(
    """

    WITH order_products_financial AS (

    SELECT

        op.order_id,

        COUNT(op.product_id) AS items_count,

        COUNT(DISTINCT op.product_id) AS unique_products,


        SUM(pf.price) AS revenue,

        SUM(pf.cogs) AS cogs,


        SUM(
            pf.price - pf.cogs
        ) AS gross_profit


    FROM order_products__prior op


    INNER JOIN product_financials pf

        ON op.product_id = pf.product_id


    GROUP BY op.order_id

)


SELECT


    o.order_id,

    o.user_id,

    o.order_number,

    o.order_dow,

    o.order_hour_of_day,

    o.days_since_prior_order,


    -- характеристики заказа

    opf.items_count,

    opf.unique_products,


    -- финансовые показатели

    opf.revenue,

    opf.cogs,

    opf.gross_profit,


    COALESCE(
        oc.delivery_cost,
        0
    ) AS delivery_cost,


    COALESCE(
        oc.marketing_cost,
        0
    ) AS marketing_cost,


    -- прибыль после переменных расходов

    (
        opf.gross_profit
        -
        COALESCE(oc.delivery_cost,0)
        -
        COALESCE(oc.marketing_cost,0)

    ) AS contribution_profit,


    -- валовая маржинальность

    ROUND(
        opf.gross_profit
        / NULLIF(opf.revenue,0)
        * 100,
        2
    ) AS gross_margin,


    -- маржинальность после переменных расходов

    ROUND(
        (
            opf.gross_profit
            -
            COALESCE(oc.delivery_cost,0)
            -
            COALESCE(oc.marketing_cost,0)
        )
        /
        NULLIF(opf.revenue,0)
        * 100,
        2
    ) AS contribution_margin



FROM order_products_financial opf


INNER JOIN orders o

    ON opf.order_id = o.order_id


LEFT JOIN order_costs oc

    ON o.order_id = oc.order_id


WHERE o.eval_set = 'prior'

    """,
    engine
)


order_unit_economics.head()
```

## 2.1 Order P&L

```python
# Финансовая модель заказа (Order P&L)


revenue = order_unit_economics["revenue"].sum()
cogs = order_unit_economics["cogs"].sum()
gross_profit = order_unit_economics["gross_profit"].sum()
delivery = order_unit_economics["delivery_cost"].sum()
marketing = order_unit_economics["marketing_cost"].sum()
contribution = order_unit_economics["contribution_profit"].sum()

financial_pnl = pd.DataFrame({

    "Показатель":[

        "Количество заказов",

        "Общая выручка (Revenue)",

        "Средний чек заказа (AOV)",

        "Себестоимость товаров (COGS)",

        "Валовая прибыль (Gross Profit)",

        "Расходы на доставку (Delivery Cost)",

        "Маркетинговые расходы (Marketing Cost)",

        "Contribution Profit"

    ],

    "Значение":[

        order_unit_economics["order_id"].nunique(),

        revenue / 1_000_000,

        order_unit_economics["revenue"].mean(),

        cogs / 1_000_000,

        gross_profit / 1_000_000,

        delivery / 1_000_000,

        marketing / 1_000_000,

        contribution / 1_000_000

    ],

    "Доля от Revenue":[

        np.nan,

        100,

        np.nan,

        cogs / revenue * 100,

        gross_profit / revenue * 100,

        delivery / revenue * 100,

        marketing / revenue * 100,

        contribution / revenue * 100

    ]

})


# Форматирование

def format_value(row):

    metric = row["Показатель"]
    value = row["Значение"]

    if metric == "Количество заказов":

        return f"{value:,.0f}".replace(",", " ")

    elif metric == "Средний чек заказа (AOV)":

        return f"{value:,.2f} ₽".replace(",", " ")

    else:

        return f"{value:,.2f} млн ₽".replace(",", " ")


financial_pnl["Значение"] = (
    financial_pnl
    .apply(format_value, axis=1)
)


financial_pnl["Доля от Revenue"] = (
    financial_pnl["Доля от Revenue"]
    .apply(
        lambda x:
        "-"
        if pd.isna(x)
        else f"{x:.2f}%"
    )
)


display(
    financial_pnl
    .style
    .hide(axis="index")
)
```

За анализируемый период было выполнено 3,2 млн заказов с общей выручкой 9,54 млрд ₽ и средним чеком 2 967 ₽. Валовая маржинальность бизнеса составила 34,66%, при этом после учета переменных расходов на доставку и маркетинг Contribution Margin снизилась до 29,78%.

Основной вклад в снижение прибыли формируют расходы на себестоимость товаров (65,34% выручки), тогда как затраты на доставку (4,05%) и маркетинг (0,83%) относительно невелики. В целом модель показывает положительную юнит-экономику: заказы в среднем генерирует положительный вклад после покрытия основных переменных расходов.

## 2.6 Анализ чувствительности юнит-экономики

```python

scenarios = [-30, -20, -10, 0, 10, 20, 30]

sensitivity_rows = []

for driver in ["cogs", "delivery_cost", "marketing_cost"]:

    for change in scenarios:

        # Изменяем только один фактор
        scenario_cogs = order_unit_economics["cogs"].copy()
        scenario_delivery = order_unit_economics["delivery_cost"].copy()
        scenario_marketing = order_unit_economics["marketing_cost"].copy()

        if driver == "cogs":
            scenario_cogs = (
                scenario_cogs
                * (1 + change / 100)
            )

        elif driver == "delivery_cost":
            scenario_delivery = (
                scenario_delivery
                * (1 + change / 100)
            )

        elif driver == "marketing_cost":
            scenario_marketing = (
                scenario_marketing
                * (1 + change / 100)
            )

        # Пересчитываем Contribution Profit
        scenario_profit = (
            order_unit_economics["revenue"]
            - scenario_cogs
            - scenario_delivery
            - scenario_marketing
        )

        # Общая Contribution Profit
        total_profit = scenario_profit.sum()

        # Средняя прибыль на заказ
        avg_profit = scenario_profit.mean()

        # Contribution Margin
        contribution_margin = (
            total_profit
            /
            order_unit_economics["revenue"].sum()
            * 100
        )

        # Убыточные заказы
        loss_orders = (
            scenario_profit < 0
        ).sum()

        # Доля убыточных заказов
        loss_share = (
            loss_orders
            /
            len(order_unit_economics)
            * 100
        )

        sensitivity_rows.append({

            "driver": driver,

            "change": change,

            "total_profit": total_profit,

            "avg_profit": avg_profit,

            "contribution_margin": contribution_margin,

            "loss_orders": loss_orders,

            "loss_share": loss_share
        })


sensitivity = pd.DataFrame(
    sensitivity_rows
)

sensitivity
```

```python
# ==========================================
# Регулярность покупок и клиентская ценность
# ==========================================

days_analysis = (
    customer_analysis[
        customer_analysis["avg_days_between_orders"].notna()
    ]
    .groupby("avg_days_between_orders")
    .agg(
        customers=("user_id", "count"),
        avg_orders=("orders", "mean"),
        avg_order_value=("avg_order_value", "mean"),
        avg_contribution_profit=("contribution_profit", "mean")
    )
    .reset_index()
)


# ==========================================
# 1. Интервал → количество заказов
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    days_analysis["avg_days_between_orders"],
    days_analysis["avg_orders"],
    marker="o",
    linewidth=2
)

plt.title(
    "Зависимость количества заказов "
    "от регулярности покупок",
    fontsize=14
)

plt.xlabel("Средний интервал между заказами, дней")
plt.ylabel("Среднее количество заказов")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ==========================================
# 2. Интервал → средний чек
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    days_analysis["avg_days_between_orders"],
    days_analysis["avg_order_value"],
    marker="o",
    linewidth=2
)

plt.title(
    "Зависимость среднего чека "
    "от регулярности покупок",
    fontsize=14
)

plt.xlabel("Средний интервал между заказами, дней")
plt.ylabel("Средний чек, ₽")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ==========================================
# 3. Интервал → Contribution Profit
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    days_analysis["avg_days_between_orders"],
    days_analysis["avg_contribution_profit"],
    marker="o",
    linewidth=2
)

plt.title(
    "Зависимость Contribution Profit клиента "
    "от регулярности покупок",
    fontsize=14
)

plt.xlabel("Средний интервал между заказами, дней")
plt.ylabel("Средний Contribution Profit, ₽")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Клиенты, которые покупают чаще, в среднем совершают больше заказов и приносят больше Revenue и Contribution Profit. При среднем интервале между заказами более 5 дней показатели начинают снижаться.


## 3.1 Анализ удержания клиентов по глубине заказов



```python
# ==========================================
# Retention по глубине заказов
# ==========================================

retention = (
    order_unit_economics[
        ["user_id", "order_number"]
    ]
    .drop_duplicates()
    .groupby("order_number")["user_id"]
    .nunique()
    .reset_index(name="customers")
    .sort_values("order_number")
)

# Доля клиентов, дошедших до каждого заказа,
# относительно первого заказа
first_order_customers = retention.loc[
    retention["order_number"] == 1,
    "customers"
].iloc[0]

retention["retention_rate"] = (
    retention["customers"] / first_order_customers
)

# ==========================================
# График
# ==========================================

plt.figure(figsize=(14, 6))

plt.plot(
    retention["order_number"],
    retention["retention_rate"] * 100,
    marker="o",
    markersize=3,
    linewidth=2
)

plt.title(
    "Retention по глубине заказов",
    fontsize=15
)

plt.xlabel("Номер заказа")
plt.ylabel("Клиенты, дошедшие до заказа, %")

plt.xticks(range(1, 100, 5))

plt.yticks(range(0, 101, 10))

plt.ylim(0, 105)

plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# Темп сокращения клиентской базы
# ==========================================

retention_change = retention.copy()

# Сокращение клиентской базы относительно предыдущего заказа
retention_change["churn_rate"] = (
    -retention_change["customers"].pct_change() * 100
)

plot_data = retention_change[
    retention_change["order_number"] >= 4
].copy()

plt.figure(figsize=(15, 6))

plt.plot(
    plot_data["order_number"],
    plot_data["churn_rate"],
    marker="o",
    markersize=3,
    linewidth=1.8
)

plt.title(
    "Темп сокращения клиентской базы "
    "при переходе к следующему заказу",
    fontsize=15
)

plt.xlabel("Номер заказа")
plt.ylabel("Сокращение клиентской базы, %")

plt.xticks(range(5, 100, 5))

plt.yticks(
    range(0, 13, 2),
    [f"{x}%" for x in range(0, 13, 2)]
)

plt.ylim(0, 12)

plt.grid(
    alpha=0.25
)

plt.tight_layout()
plt.show()
```

Темп сокращения клиентской базы постепенно снижается с увеличением глубины заказов. На ранних этапах он уменьшается примерно с 12% до 8%, затем постепенно снижается до 5–6% к 20-му заказу и до 3–4% после 65-го заказа. После этого наблюдаются небольшие колебания вокруг этого уровня.

Это показывает, что по мере увеличения глубины покупок клиентская база становится более устойчивой: клиенты, которые уже совершили большое количество заказов, с большей вероятностью продолжают покупать. При этом это не классический временной retention — результат частично связан с тем, что до высокой глубины доходят наиболее лояльные клиенты.

## 3.2 Экономика клиентов в зависимости от глубины покупок

```python
# ==========================================
# Количество клиентов на каждой глубине
# ==========================================

plt.figure(figsize=(14, 5))

plt.bar(
    orders_analysis["orders"],
    orders_analysis["customers"]
)

plt.title(
    "Количество клиентов в зависимости от глубины покупок",
    fontsize=14
)

plt.xlabel("Количество заказов клиента")
plt.ylabel("Количество клиентов")

plt.xticks(range(1, 100, 5))

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# Глубина покупок клиента
# ==========================================

customer_depth = (
    order_unit_economics
    .groupby("user_id")
    .agg(
        last_order=("order_number", "max")
    )
    .reset_index()
)

customer_depth["last_order"].describe(
    percentiles=[0.25, 0.5, 0.75, 0.90, 0.95, 0.99]
)


# ==========================================
# Сегментация по глубине покупок
# ==========================================

customer_depth["depth_segment"] = pd.cut(
    customer_depth["last_order"],
    bins=[2, 5, 9, 19, 37, 51, 99],
    labels=[
        "3–5 заказов",
        "6–9 заказов",
        "10–19 заказов",
        "20–37 заказов",
        "38–51 заказ",
        "52–99 заказов"
    ]
)


# ==========================================
# Экономика клиентов по глубине
# ==========================================

depth_analysis = (
    customer_analysis
    .merge(
        customer_depth[
            ["user_id", "depth_segment"]
        ],
        on="user_id",
        how="left"
    )
    .groupby(
        "depth_segment",
        observed=False
    )
    .agg(
        customers=("user_id", "count"),

        avg_orders=(
            "orders",
            "mean"
        ),

        avg_contribution_profit=(
            "contribution_profit",
            "mean"
        ),

        avg_profit_per_order=(
            "avg_profit_per_order",
            "mean"
        )
    )
    .reset_index()
)


# ==========================================
# 1. Заказов на клиента
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    depth_analysis["depth_segment"].astype(str),
    depth_analysis["avg_orders"],
    marker="o",
    linewidth=2
)

plt.title(
    "Количество заказов на клиента "
    "в зависимости от глубины покупок",
    fontsize=14
)

plt.xlabel("Глубина покупок")
plt.ylabel("Заказов на клиента")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ==========================================
# 2. Contribution Profit на клиента
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    depth_analysis["depth_segment"].astype(str),
    depth_analysis["avg_contribution_profit"],
    marker="o",
    linewidth=2
)

plt.title(
    "Contribution Profit на клиента "
    "в зависимости от глубины покупок",
    fontsize=14
)

plt.xlabel("Глубина покупок")
plt.ylabel("Contribution Profit на клиента, ₽")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ==========================================
# 3. Прибыль на заказ
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    depth_analysis["depth_segment"].astype(str),
    depth_analysis["avg_profit_per_order"],
    marker="o",
    linewidth=2
)

plt.title(
    "Прибыль на заказ "
    "в зависимости от глубины покупок",
    fontsize=14
)

plt.xlabel("Глубина покупок")
plt.ylabel("Прибыль на заказ, ₽")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# Экономика заказа по глубине покупок
# ==========================================

depth_economics = (
    order_unit_economics
    .merge(
        customer_depth[
            ["user_id", "depth_segment"]
        ],
        on="user_id",
        how="left"
    )
)


depth_economics_analysis = (
    depth_economics
    .groupby(
        "depth_segment",
        observed=False
    )
    .agg(
        orders=("order_id", "count"),

        avg_revenue=("revenue", "mean"),

        avg_cogs=("cogs", "mean"),

        avg_gross_profit=(
            "gross_profit",
            "mean"
        ),

        avg_delivery_cost=(
            "delivery_cost",
            "mean"
        ),

        avg_marketing_cost=(
            "marketing_cost",
            "mean"
        ),

        avg_contribution_profit=(
            "contribution_profit",
            "mean"
        ),

        avg_items=(
            "items_count",
            "mean"
        )
    )
    .reset_index()
)


# ==========================================
# Оформление
# ==========================================

depth_economics_display = (
    depth_economics_analysis
    .rename(columns={
        "depth_segment": "Глубина покупок",
        "orders": "Заказы",
        "avg_revenue": "Revenue на заказ",
        "avg_cogs": "COGS на заказ",
        "avg_gross_profit": "Gross Profit на заказ",
        "avg_delivery_cost": "Доставка на заказ",
        "avg_marketing_cost": "Маркетинг на заказ",
        "avg_contribution_profit": "Contribution Profit на заказ",
        "avg_items": "Товаров в заказе"
    })
)


display(
    depth_economics_display.style
    .format({
        "Заказы": "{:,.0f}",
        "Revenue на заказ": "{:,.0f} ₽",
        "COGS на заказ": "{:,.0f} ₽",
        "Gross Profit на заказ": "{:,.0f} ₽",
        "Доставка на заказ": "{:,.0f} ₽",
        "Маркетинг на заказ": "{:,.0f} ₽",
        "Contribution Profit на заказ": "{:,.0f} ₽",
        "Товаров в заказе": "{:.1f}"
    })
    .background_gradient(
        subset=[
            "Revenue на заказ",
            "Gross Profit на заказ",
            "Contribution Profit на заказ",
            "Товаров в заказе"
        ],
        cmap="Greens"
    )
    .set_properties(**{
        "text-align": "center",
        "padding": "8px"
    })
)
```

С ростом глубины покупок экономика заказа постепенно улучшается до сегмента 38–51 заказ: Contribution Profit на заказ увеличивается с 845 ₽ до 937 ₽ (+11%). Основной фактор — рост Revenue и размера корзины: количество товаров увеличивается с 9,7 до 10,6, а Revenue — с 2 853 до 3 124 ₽.

При этом Delivery Cost и Marketing Cost практически не меняются, поэтому дополнительная выручка преимущественно трансформируется в Gross Profit и Contribution Profit.

Однако в сегменте 52–99 заказов экономика заказа возвращается к уровню клиентов с 3–5 заказами: Contribution Profit снижается до 845 ₽, а корзина — до 9,7 товара.

Вывод: глубина покупок сама по себе не улучшает экономику каждого заказа. Наиболее эффективны клиенты с 38–51 заказом, тогда как у самых глубоких клиентов наблюдается снижение размера корзины и прибыльности заказа. При этом их высокая совокупная ценность сохраняется прежде всего за счёт значительно большей частоты покупок.

## 3.3 Сегментация клиентов

```python
# ==========================================
# 3.3. Сегментация клиентов по ценности
# ==========================================

profit_segments = (
    customer_analysis
    .assign(
        segment=pd.qcut(
            customer_analysis["contribution_profit"],
            q=4,
            labels=[
                "Низкая ценность",
                "Средняя ценность",
                "Высокая ценность",
                "Очень высокая ценность"
            ]
        )
    )
    .groupby("segment", observed=False)
    .agg(
        customers=("user_id", "count"),
        avg_orders=("orders", "mean"),
        avg_contribution_profit=("contribution_profit", "mean"),
        avg_order_value=("avg_order_value", "mean"),
        avg_profit_per_order=("avg_profit_per_order", "mean"),
        avg_items=("avg_items", "mean"),
        avg_margin=("avg_margin", "mean"),
        avg_days_between_orders=("avg_days_between_orders", "mean"),
        total_contribution_profit=("contribution_profit", "sum")
    )
    .reset_index()
)

# Доля Contribution Profit

profit_segments["profit_share"] = (
    profit_segments["total_contribution_profit"]
    / profit_segments["total_contribution_profit"].sum()
)

# ==========================================
# Вывод
# ==========================================

display(
    profit_segments
    .rename(columns={
        "segment": "Сегмент",
        "customers": "Клиенты",
        "avg_orders": "Заказов на клиента",
        "avg_contribution_profit": "Contribution Profit на клиента",
        "avg_order_value": "Средний чек",
        "avg_profit_per_order": "Прибыль на заказ",
        "avg_items": "Товаров в заказе",
        "avg_margin": "Маржинальность",
        "avg_days_between_orders": "Интервал между заказами, дней",
        "total_contribution_profit": "Contribution Profit сегмента",
        "profit_share": "Доля Contribution Profit"
    })
    .style
    .format({
        "Клиенты": "{:,.0f}",
        "Заказов на клиента": "{:,.1f}",
        "Contribution Profit на клиента": "{:,.0f} ₽",
        "Средний чек": "{:,.0f} ₽",
        "Прибыль на заказ": "{:,.0f} ₽",
        "Товаров в заказе": "{:,.1f}",
        "Маржинальность": "{:.1f}%",
        "Интервал между заказами, дней": "{:,.1f}",
        "Contribution Profit сегмента": "{:,.0f} ₽",
        "Доля Contribution Profit": "{:.1%}"
    })
    .background_gradient(
        subset=[
            "Заказов на клиента",
            "Contribution Profit на клиента",
            "Средний чек",
            "Прибыль на заказ",
            "Товаров в заказе",
            "Маржинальность",
            "Доля Contribution Profit"
        ],
        cmap="Greens"
    )
    .background_gradient(
        subset=["Интервал между заказами, дней"],
        cmap="Oranges"
    )
    .set_properties(**{
        "text-align": "center",
        "padding": "8px"
    })
)
```

```python
# ==========================================
# Клиентская ценность по сегментам
# ==========================================

plot_data = profit_segments_display.copy()

plt.figure(figsize=(14, 6))

plt.plot(
    plot_data["Сегмент"],
    plot_data["Contribution Profit на клиента"],
    marker="o",
    linewidth=2.5
)

plt.title(
    "Клиентская ценность по сегментам",
    fontsize=15
)

plt.xlabel("Сегмент")
plt.ylabel("Contribution Profit на клиента, ₽")

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()
plt.show()
```

Клиентская ценность существенно различается между сегментами: 25% клиентов с очень высокой ценностью формируют 68,9% общего Contribution Profit. Их преимущество обусловлено одновременно более высокой частотой покупок (34,1 заказа), крупной корзиной (14 товаров) и более высокой прибылью на заказ (1 292 ₽). При этом интервал между покупками сокращается до 10,3 дня против 18,6 дня у клиентов низкой ценности.

Вывод: основной драйвер накопленной клиентской ценности — сочетание частоты покупок и экономики заказа. Поэтому приоритетная зона роста — увеличение повторных покупок и перевод клиентов в более высокие ценностные сегменты.

Клиентская ценность формируется прежде всего сочетанием частоты покупок, регулярности и экономики заказа. По мере увеличения глубины покупок клиентская база становится более устойчивой, а клиенты с высокой глубиной заказа демонстрируют более высокую накопленную прибыльность.

При этом экономика отдельного заказа растёт не линейно: максимальный Contribution Profit на заказ наблюдается у клиентов с 38–51 заказом и достигает 937 ₽. Дальнейшее увеличение глубины покупок не приводит к улучшению экономики заказа — в сегменте 52–99 заказов показатели возвращаются примерно к уровню клиентов с 3–5 заказами. Следовательно, высокая ценность наиболее лояльных клиентов формируется преимущественно за счёт высокой частоты покупок, а не постоянного роста прибыльности каждого заказа.

Сегментация подтверждает высокую концентрацию клиентской прибыли: 25% клиентов с очень высокой ценностью формируют 68,9% общего Contribution Profit. Они совершают в среднем 34,1 заказа, имеют более крупную корзину и более короткий интервал между покупками.

Бизнес-вывод: ключевой драйвер долгосрочной клиентской ценности — сочетание частоты покупок и прибыльности заказа. Основная точка роста — повышение повторной активности клиентов и перевод клиентов в более высокие ценностные сегменты при сохранении положительной экономики заказа.
