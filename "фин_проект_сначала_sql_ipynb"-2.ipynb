```python
# Проверка NULL

def check_nulls(table_name):

    # Получаем список столбцов таблицы
    columns = pd.read_sql(
        f"PRAGMA table_info({table_name});",
        engine
    )["name"].tolist()


    # Получаем количество строк
    rows_count = pd.read_sql(
        f"SELECT COUNT(*) AS cnt FROM {table_name};",
        engine
    ).loc[0, "cnt"]


    # Формируем один SQL-запрос
    null_checks = []

    for column in columns:

        null_checks.append(
            f"""
            SUM(
                CASE
                    WHEN {column} IS NULL THEN 1
                    ELSE 0
                END
            ) AS {column}
            """
        )


    sql = f"""
    SELECT
        {",".join(null_checks)}
    FROM {table_name};
    """


    # Выполняем один проход по таблице
    result = pd.read_sql(sql, engine)


    # Переводим широкий формат в удобный
    result = result.T.reset_index()

    result.columns = [
        "column",
        "null_count"
    ]


    result["table"] = table_name

    result["null_percent"] = (
        result["null_count"]
        /
        rows_count
        *
        100
    ).round(2)


    return result[
        [
            "table",
            "column",
            "null_count",
            "null_percent"
        ]
    ]
```

```python
nulls_check = pd.concat(
    [
        check_nulls(table)
        for table in table_list
    ],
    ignore_index=True
)

nulls_check
```

```python
# Дубликаты

def check_duplicates(table_name, column):

    # Ищем только значения ключа,
    # которые встречаются больше одного раза.
    # Это позволяет не считать все DISTINCT значения.

    sql = f"""
    SELECT
        COUNT(*) AS duplicate_rows
    FROM
    (
        SELECT
            {column}
        FROM {table_name}
        GROUP BY {column}
        HAVING COUNT(*) > 1
    );
    """


    result = pd.read_sql(sql, engine)


    # Получаем общее количество строк
    sql_total = f"""
    SELECT COUNT(*) AS total_rows
    FROM {table_name};
    """

    total_rows = pd.read_sql(
        sql_total,
        engine
    ).loc[0, "total_rows"]


    result["table"] = table_name
    result["column"] = column

    result["total_rows"] = total_rows

    result["duplicate_percent"] = (
        result["duplicate_rows"]
        /
        total_rows
        *
        100
    ).round(2)


    return result[
        [
            "table",
            "column",
            "total_rows",
            "duplicate_rows",
            "duplicate_percent"
        ]
    ]
```

#  1. Анализ продаж и ассортимента продуктов питания

```python
product_sales_analysis = pd.read_sql(
    """
    WITH product_sales AS (

        -- продажи по каждому товару

        SELECT

            product_id,

            COUNT(DISTINCT order_id) AS orders_count,

            COUNT(product_id) AS units_sold

        FROM order_products__prior

        GROUP BY product_id

    )


    SELECT

        p.product_id,

        p.product_name,

        d.department,


        -- продажи

        COALESCE(ps.orders_count,0) AS orders_count,

        COALESCE(ps.units_sold,0) AS units_sold,


        -- финансовые показатели

        COALESCE(
            ps.units_sold * pf.price,
            0
        ) AS revenue,


        COALESCE(
            ps.units_sold * pf.cogs,
            0
        ) AS cogs,


        COALESCE(
            ps.units_sold * (pf.price - pf.cogs),
            0
        ) AS gross_profit


    FROM products p


    LEFT JOIN product_sales ps

        ON p.product_id = ps.product_id


    LEFT JOIN departments d

        ON p.department_id = d.department_id


    LEFT JOIN product_financials pf

        ON p.product_id = pf.product_id

    """,
    engine
)


product_sales_analysis.head()
```

```python
# Маржинальность товара

product_sales_analysis["margin"] = np.where(
    product_sales_analysis["revenue"] > 0,
    (
        product_sales_analysis["gross_profit"]
        /
        product_sales_analysis["revenue"]
        * 100
    ).round(2),
    0
)
```

```python
# Статус товара

product_sales_analysis["sales_status"] = np.where(
    product_sales_analysis["units_sold"] > 0,
    "Продается",
    "Нет продаж"
)
```

## 1.1. Структура и эффективность ассортимента

```python
# Общее количество товаров

print(
    "Количество товаров:",
    product_sales_analysis["product_id"].nunique()
)
```

```python
# Общее количество категорий

print(
    "Количество категорий  товаров:",
    product_sales_analysis["department"].nunique()
)

print(
    "Категории товаров:",
    product_sales_analysis["department"].unique()
)
```

```python
# Анализ эффективности категорий

category_efficiency = (
    product_sales_analysis
    .groupby("department")
    .agg(
        products=("product_id","count"),
        units_sold=("units_sold","sum"),
        revenue=("revenue","sum"),
        profit=("gross_profit","sum")
    )
)


# Доля категории в ассортименте

category_efficiency["assortment_share"] = (
    category_efficiency["products"]
    /
    category_efficiency["products"].sum()
    * 100
).round(2)


# Доля выручки

category_efficiency["revenue_share"] = (
    category_efficiency["revenue"]
    /
    category_efficiency["revenue"].sum()
    * 100
).round(2)


# Доля прибыли

category_efficiency["profit_share"] = (
    category_efficiency["profit"]
    /
    category_efficiency["profit"].sum()
    * 100
).round(2)

# Эффективность ассортимента:
# сколько доли прибыли приносит 1% ассортимента

category_efficiency["assortment_efficiency"] = (
    category_efficiency["profit_share"]
    /
    category_efficiency["assortment_share"]
).round(2)


# Маржинальность

category_efficiency["margin"] = (
    category_efficiency["profit"]
    /
    category_efficiency["revenue"]
    * 100
).round(2)


# Перевод в миллионы

category_efficiency["revenue_mln"] = (
    category_efficiency["revenue"] / 1_000_000
).round(2)


category_efficiency["profit_mln"] = (
    category_efficiency["profit"] / 1_000_000
).round(2)
```

```python
# копия исходной таблицы
abc_profit = product_sales_analysis.copy()


# сортируем товары по валовой прибыли
abc_profit = (
    abc_profit
    .sort_values(
        "gross_profit",
        ascending=False
    )
)


# доля товара в общей прибыли

abc_profit["profit_share"] = (
    abc_profit["gross_profit"]
    /
    abc_profit["gross_profit"].sum()
)


# накопительная доля прибыли

abc_profit["cumulative_profit_share"] = (
    abc_profit["profit_share"]
    .cumsum()
)


# присваиваем ABC-сегмент

abc_profit["ABC_segment"] = np.where(
    abc_profit["cumulative_profit_share"] <= 0.8,
    "A",
    np.where(
        abc_profit["cumulative_profit_share"] <= 0.95,
        "B",
        "C"
    )
)


# товары без продаж выделяем отдельно

abc_profit.loc[
    abc_profit["units_sold"] == 0,
    "ABC_segment"
] = "Нет продаж"


# Проверяем распределение товаров

abc_profit["ABC_segment"].value_counts()
```

#  2. Юнит-экономика заказа

```python
# Объединяем данные о заказах, товарах,
# финансовых показателях товаров и расходах по заказам

order_unit_economics = pd.read_sql(
    """

    WITH order_products_financial AS (

    SELECT

        op.order_id,

        COUNT(op.product_id) AS items_count,

        COUNT(DISTINCT op.product_id) AS unique_products,


        SUM(pf.price) AS revenue,

        SUM(pf.cogs) AS cogs,


        SUM(
            pf.price - pf.cogs
        ) AS gross_profit


    FROM order_products__prior op


    INNER JOIN product_financials pf

        ON op.product_id = pf.product_id


    GROUP BY op.order_id

)


SELECT


    o.order_id,

    o.user_id,

    o.order_number,

    o.order_dow,

    o.order_hour_of_day,

    o.days_since_prior_order,


    -- характеристики заказа

    opf.items_count,

    opf.unique_products,


    -- финансовые показатели

    opf.revenue,

    opf.cogs,

    opf.gross_profit,


    COALESCE(
        oc.delivery_cost,
        0
    ) AS delivery_cost,


    COALESCE(
        oc.marketing_cost,
        0
    ) AS marketing_cost,


    -- прибыль после переменных расходов

    (
        opf.gross_profit
        -
        COALESCE(oc.delivery_cost,0)
        -
        COALESCE(oc.marketing_cost,0)

    ) AS contribution_profit,


    -- валовая маржинальность

    ROUND(
        opf.gross_profit
        / NULLIF(opf.revenue,0)
        * 100,
        2
    ) AS gross_margin,


    -- маржинальность после переменных расходов

    ROUND(
        (
            opf.gross_profit
            -
            COALESCE(oc.delivery_cost,0)
            -
            COALESCE(oc.marketing_cost,0)
        )
        /
        NULLIF(opf.revenue,0)
        * 100,
        2
    ) AS contribution_margin



FROM order_products_financial opf


INNER JOIN orders o

    ON opf.order_id = o.order_id


LEFT JOIN order_costs oc

    ON o.order_id = oc.order_id


WHERE o.eval_set = 'prior'

    """,
    engine
)


order_unit_economics.head()
```

## 2.1 Order P&L

```python
# Финансовая модель заказа (Order P&L)


revenue = order_unit_economics["revenue"].sum()
cogs = order_unit_economics["cogs"].sum()
gross_profit = order_unit_economics["gross_profit"].sum()
delivery = order_unit_economics["delivery_cost"].sum()
marketing = order_unit_economics["marketing_cost"].sum()
contribution = order_unit_economics["contribution_profit"].sum()

financial_pnl = pd.DataFrame({

    "Показатель":[

        "Количество заказов",

        "Общая выручка (Revenue)",

        "Средний чек заказа (AOV)",

        "Себестоимость товаров (COGS)",

        "Валовая прибыль (Gross Profit)",

        "Расходы на доставку (Delivery Cost)",

        "Маркетинговые расходы (Marketing Cost)",

        "Contribution Profit"

    ],

    "Значение":[

        order_unit_economics["order_id"].nunique(),

        revenue / 1_000_000,

        order_unit_economics["revenue"].mean(),

        cogs / 1_000_000,

        gross_profit / 1_000_000,

        delivery / 1_000_000,

        marketing / 1_000_000,

        contribution / 1_000_000

    ],

    "Доля от Revenue":[

        np.nan,

        100,

        np.nan,

        cogs / revenue * 100,

        gross_profit / revenue * 100,

        delivery / revenue * 100,

        marketing / revenue * 100,

        contribution / revenue * 100

    ]

})


# Форматирование

def format_value(row):

    metric = row["Показатель"]
    value = row["Значение"]

    if metric == "Количество заказов":

        return f"{value:,.0f}".replace(",", " ")

    elif metric == "Средний чек заказа (AOV)":

        return f"{value:,.2f} ₽".replace(",", " ")

    else:

        return f"{value:,.2f} млн ₽".replace(",", " ")


financial_pnl["Значение"] = (
    financial_pnl
    .apply(format_value, axis=1)
)


financial_pnl["Доля от Revenue"] = (
    financial_pnl["Доля от Revenue"]
    .apply(
        lambda x:
        "-"
        if pd.isna(x)
        else f"{x:.2f}%"
    )
)


display(
    financial_pnl
    .style
    .hide(axis="index")
)
```

За анализируемый период было выполнено 3,2 млн заказов с общей выручкой 9,54 млрд ₽ и средним чеком 2 967 ₽. Валовая маржинальность бизнеса составила 34,66%, при этом после учета переменных расходов на доставку и маркетинг Contribution Margin снизилась до 29,78%.

Основной вклад в снижение прибыли формируют расходы на себестоимость товаров (65,34% выручки), тогда как затраты на доставку (4,05%) и маркетинг (0,83%) относительно невелики. В целом модель показывает положительную юнит-экономику: заказы в среднем генерирует положительный вклад после покрытия основных переменных расходов.

## 2.6 Анализ чувствительности юнит-экономики

```python

scenarios = [-30, -20, -10, 0, 10, 20, 30]

sensitivity_rows = []

for driver in ["cogs", "delivery_cost", "marketing_cost"]:

    for change in scenarios:

        # Изменяем только один фактор
        scenario_cogs = order_unit_economics["cogs"].copy()
        scenario_delivery = order_unit_economics["delivery_cost"].copy()
        scenario_marketing = order_unit_economics["marketing_cost"].copy()

        if driver == "cogs":
            scenario_cogs = (
                scenario_cogs
                * (1 + change / 100)
            )

        elif driver == "delivery_cost":
            scenario_delivery = (
                scenario_delivery
                * (1 + change / 100)
            )

        elif driver == "marketing_cost":
            scenario_marketing = (
                scenario_marketing
                * (1 + change / 100)
            )

        # Пересчитываем Contribution Profit
        scenario_profit = (
            order_unit_economics["revenue"]
            - scenario_cogs
            - scenario_delivery
            - scenario_marketing
        )

        # Общая Contribution Profit
        total_profit = scenario_profit.sum()

        # Средняя прибыль на заказ
        avg_profit = scenario_profit.mean()

        # Contribution Margin
        contribution_margin = (
            total_profit
            /
            order_unit_economics["revenue"].sum()
            * 100
        )

        # Убыточные заказы
        loss_orders = (
            scenario_profit < 0
        ).sum()

        # Доля убыточных заказов
        loss_share = (
            loss_orders
            /
            len(order_unit_economics)
            * 100
        )

        sensitivity_rows.append({

            "driver": driver,

            "change": change,

            "total_profit": total_profit,

            "avg_profit": avg_profit,

            "contribution_margin": contribution_margin,

            "loss_orders": loss_orders,

            "loss_share": loss_share
        })


sensitivity = pd.DataFrame(
    sensitivity_rows
)

sensitivity
```

При этом доставка остается важным фактором на уровне отдельных заказов, особенно для небольших корзин, где ее доля в выручке значительно выше.

```python
# ==========================================
# 3.4. RETENTION ПО НОМЕРУ ЗАКАЗА
# ==========================================

# ==========================================
# 1. Данные для retention
# ==========================================

retention_data = (
    order_unit_economics[
        ["user_id", "order_number"]
    ]
    .drop_duplicates()
)


# ==========================================
# 2. Количество клиентов,
#    совершивших каждый по счёту заказ
# ==========================================

retention = (
    retention_data
    .groupby("order_number")
    .agg(
        customers=("user_id", "nunique")
    )
    .reset_index()
    .sort_values("order_number")
)


# ==========================================
# 3. Размер клиентской базы
#    на первом заказе
# ==========================================

first_order_customers = retention.loc[
    retention["order_number"] == 1,
    "customers"
].iloc[0]


# ==========================================
# 4. Retention относительно первого заказа
# ==========================================

retention["retention_rate"] = (
    retention["customers"]
    / first_order_customers
)


# ==========================================
# 5. Retention относительно
#    предыдущего заказа
# ==========================================

retention["previous_customers"] = (
    retention["customers"].shift(1)
)

retention["step_retention"] = (
    retention["customers"]
    / retention["previous_customers"]
)


# ==========================================
# 6. Retention table
# ==========================================

retention_display = (
    retention[
        [
            "order_number",
            "customers",
            "retention_rate",
            "step_retention"
        ]
    ]
    .rename(
        columns={
            "order_number": "Номер заказа",
            "customers": "Клиенты",
            "retention_rate": "Retention от 1-го заказа",
            "step_retention": "Retention от предыдущего заказа"
        }
    )
)


display(
    retention_display.style
    .format({
        "Клиенты": "{:,.0f}",
        "Retention от 1-го заказа": "{:.1%}",
        "Retention от предыдущего заказа": "{:.1%}"
    })
)
```

```python
# ==========================================
# 7. Retention curve
# ==========================================

plt.figure(figsize=(12, 6))

plt.plot(
    retention["order_number"],
    retention["retention_rate"] * 100,
    marker="o",
    linewidth=2
)

plt.title(
    "Retention по номеру заказа",
    fontsize=14
)

plt.xlabel("Номер заказа")
plt.ylabel("Доля клиентов от первого заказа, %")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
# ==========================================
# Сокращение клиентской базы
# ==========================================

retention_change = retention.copy()

# Сколько клиентов потеряли
retention_change["customers_lost"] = (
    retention_change["customers"].shift(1)
    - retention_change["customers"]
)

# На сколько процентов сократилась база
retention_change["change_vs_previous"] = (
    retention_change["customers"]
    .pct_change() * 100
)

# Сколько процентов клиентов перешло
# на следующий заказ
retention_change["step_retention"] = (
    retention_change["customers"]
    / retention_change["customers"].shift(1)
)

display(
    retention_change.style.format({
        "customers": "{:,.0f}",
        "customers_lost": "{:,.0f}",
        "change_vs_previous": "{:+.1f}%",
        "step_retention": "{:.1%}"
    })
)
```

```python
# ==========================================
# Глубина покупок → клиентская ценность
# ==========================================

# Максимальный номер заказа каждого клиента
customer_depth = (
    order_unit_economics
    .groupby("user_id")
    .agg(
        last_order=("order_number", "max")
    )
    .reset_index()
)

# Добавляем глубину покупок к клиентской аналитике
customer_value_depth = customer_analysis.merge(
    customer_depth,
    on="user_id",
    how="left"
)

# ==========================================
# Сегментация по глубине покупок
# ==========================================

customer_value_depth["depth_segment"] = pd.cut(
    customer_value_depth["last_order"],
    bins=[3, 5, 10, 20, 40, 65, 99],
    labels=[
        "4–5 заказов",
        "6–10 заказов",
        "11–20 заказов",
        "21–40 заказов",
        "41–65 заказов",
        "66–99 заказов"
    ],
    include_lowest=True
)

# ==========================================
# Экономика сегментов
# ==========================================

depth_analysis = (
    customer_value_depth
    .groupby("depth_segment", observed=False)
    .agg(
        customers=("user_id", "count"),

        avg_orders=("orders", "mean"),

        avg_revenue=("revenue", "mean"),

        avg_contribution_profit=(
            "contribution_profit",
            "mean"
        ),

        avg_order_value=(
            "avg_order_value",
            "mean"
        ),

        avg_profit_per_order=(
            "avg_profit_per_order",
            "mean"
        ),

        avg_items=("avg_items", "mean"),

        avg_margin=("avg_margin", "mean")
    )
    .reset_index()
)

depth_analysis
```

```python
# ==========================================
# Вклад сегмента в общий Contribution Profit
# ==========================================

depth_analysis["total_contribution_profit"] = (
    depth_analysis["customers"]
    * depth_analysis["avg_contribution_profit"]
)

total_cp = depth_analysis[
    "total_contribution_profit"
].sum()

depth_analysis["contribution_profit_share"] = (
    depth_analysis["total_contribution_profit"]
    / total_cp
)

display(
    depth_analysis.style.format({
        "customers": "{:,.0f}",
        "avg_orders": "{:,.1f}",
        "avg_revenue": "{:,.0f} ₽",
        "avg_contribution_profit": "{:,.0f} ₽",
        "avg_order_value": "{:,.0f} ₽",
        "avg_profit_per_order": "{:,.0f} ₽",
        "avg_items": "{:,.1f}",
        "avg_margin": "{:.1%}",
        "total_contribution_profit": "{:,.0f} ₽",
        "contribution_profit_share": "{:.1%}"
    })
)
```

```python
# ==========================================
# Почему падает прибыль на заказ
# после 65 заказов?
# ==========================================

depth_economics = (
    order_unit_economics
    .merge(
        customer_depth,
        on="user_id",
        how="left"
    )
)

depth_economics["depth_segment"] = pd.cut(
    depth_economics["last_order"],
    bins=[3, 5, 10, 20, 40, 65, 99],
    labels=[
        "4–5 заказов",
        "6–10 заказов",
        "11–20 заказов",
        "21–40 заказов",
        "41–65 заказов",
        "66–99 заказов"
    ],
    include_lowest=True
)

depth_economics_analysis = (
    depth_economics
    .groupby("depth_segment", observed=False)
    .agg(
        orders=("order_id", "count"),

        avg_revenue=("revenue", "mean"),

        avg_cogs=("cogs", "mean"),

        avg_gross_profit=("gross_profit", "mean"),

        avg_delivery_cost=("delivery_cost", "mean"),

        avg_marketing_cost=("marketing_cost", "mean"),

        avg_contribution_profit=(
            "contribution_profit",
            "mean"
        ),

        avg_items=("items_count", "mean"),

        avg_gross_margin=("gross_margin", "mean"),

        avg_contribution_margin=(
            "contribution_margin",
            "mean"
        )
    )
    .reset_index()
)

display(
    depth_economics_analysis.style.format({
        "orders": "{:,.0f}",
        "avg_revenue": "{:,.0f} ₽",
        "avg_cogs": "{:,.0f} ₽",
        "avg_gross_profit": "{:,.0f} ₽",
        "avg_delivery_cost": "{:,.0f} ₽",
        "avg_marketing_cost": "{:,.0f} ₽",
        "avg_contribution_profit": "{:,.0f} ₽",
        "avg_items": "{:,.1f}",
        "avg_gross_margin": "{:.1%}",
        "avg_contribution_margin": "{:.1%}"
    })
)
```

После 65 заказов наблюдается перелом: средний чек снижается на 12%, размер корзины — на 12,3%, что приводит к снижению Contribution Profit на заказ на 13,8%. При этом расходы на доставку и маркетинг практически не меняются, поэтому снижение прибыльности связано прежде всего с уменьшением размера корзины, а не с ростом переменных расходов.

```python
# ==========================================
# Структура корзины по глубине покупок
# ==========================================

depth_basket = (
    order_unit_economics
    .merge(
        customer_depth,
        on="user_id",
        how="left"
    )
)

depth_basket["depth_segment"] = pd.cut(
    depth_basket["last_order"],
    bins=[3, 5, 10, 20, 40, 65, 99],
    labels=[
        "4–5 заказов",
        "6–10 заказов",
        "11–20 заказов",
        "21–40 заказов",
        "41–65 заказов",
        "66–99 заказов"
    ],
    include_lowest=True
)

basket_structure = (
    depth_basket
    .groupby("depth_segment", observed=False)
    .agg(
        orders=("order_id", "count"),
        avg_items=("items_count", "mean"),
        avg_revenue=("revenue", "mean"),
        avg_contribution_profit=(
            "contribution_profit",
            "mean"
        )
    )
    .reset_index()
)

display(
    basket_structure.style.format({
        "orders": "{:,.0f}",
        "avg_items": "{:.1f}",
        "avg_revenue": "{:,.0f} ₽",
        "avg_contribution_profit": "{:,.0f} ₽"
    })
)
```

```python
# ==========================================
# Экономика заказа по глубине покупок
# ==========================================

depth_economics_analysis["delivery_share"] = (
    depth_economics_analysis["avg_delivery_cost"]
    / depth_economics_analysis["avg_revenue"]
)

depth_economics_analysis["marketing_share"] = (
    depth_economics_analysis["avg_marketing_cost"]
    / depth_economics_analysis["avg_revenue"]
)

display(
    depth_economics_analysis[
        [
            "depth_segment",
            "avg_revenue",
            "avg_items",
            "avg_gross_profit",
            "avg_delivery_cost",
            "delivery_share",
            "avg_marketing_cost",
            "marketing_share",
            "avg_contribution_profit"
        ]
    ].style.format({
        "avg_revenue": "{:,.0f} ₽",
        "avg_items": "{:.1f}",
        "avg_gross_profit": "{:,.0f} ₽",
        "avg_delivery_cost": "{:,.0f} ₽",
        "delivery_share": "{:.1%}",
        "avg_marketing_cost": "{:,.0f} ₽",
        "marketing_share": "{:.1%}",
        "avg_contribution_profit": "{:,.0f} ₽"
    })
)
```

```python
# ==========================================
# AOV и размер корзины
# ==========================================

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(
    depth_economics_analysis["depth_segment"].astype(str),
    depth_economics_analysis["avg_revenue"],
    marker="o",
    linewidth=2,
    label="Средний чек"
)

ax1.set_xlabel("Глубина покупок")
ax1.set_ylabel("Средний чек, ₽")

ax2 = ax1.twinx()

ax2.plot(
    depth_economics_analysis["depth_segment"].astype(str),
    depth_economics_analysis["avg_items"],
    marker="o",
    linewidth=2,
    linestyle="--",
    label="Товаров в заказе"
)

ax2.set_ylabel("Товаров в заказе")

plt.title(
    "Размер корзины в зависимости от глубины покупок",
    fontsize=14
)

ax1.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Contribution Profit: 845 → 930 ₽
доля доставки снижается: 4,2% → 3,9%

То есть по мере роста глубины покупок клиенты в среднем формируют более крупные и более выгодные корзины.

После 65 заказов происходит перелом:

Главный вывод

По мере увеличения глубины покупок экономика заказа постепенно улучшается и достигает максимума в сегменте 41–65 заказов. После 65 заказов наблюдается обратная динамика: самые частотные клиенты начинают формировать более компактные корзины, вследствие чего средний чек и Contribution Profit на заказ снижаются. При этом абсолютные расходы на доставку и маркетинг практически не меняются, поэтому основным фактором снижения прибыльности является уменьшение размера корзины.

```python
# ==========================================
# LTV по глубине клиентского цикла
# Все 99 заказов
# ==========================================

# Суммируем показатели клиента
ltv_by_customer = (
    order_unit_economics
    .groupby("user_id")
    .agg(
        orders=("order_id", "count"),
        revenue=("revenue", "sum"),
        contribution_profit=("contribution_profit", "sum")
    )
    .reset_index()
)

# Разбиваем клиентов по глубине
ltv_by_depth = (
    ltv_by_customer
    .groupby("orders")
    .agg(
        customers=("user_id", "count"),
        avg_revenue_ltv=("revenue", "mean"),
        avg_contribution_ltv=("contribution_profit", "mean"),
        median_revenue_ltv=("revenue", "median"),
        median_contribution_ltv=(
            "contribution_profit",
            "median"
        )
    )
    .reset_index()
)

ltv_by_depth
```

```python
# ==========================================
# Накопленный LTV по глубине покупок
# ==========================================

plt.figure(figsize=(14, 6))

plt.plot(
    ltv_by_depth["orders"],
    ltv_by_depth["avg_contribution_ltv"],
    marker="o",
    markersize=3,
    linewidth=2,
    label="Средний Contribution Profit LTV"
)

plt.plot(
    ltv_by_depth["orders"],
    ltv_by_depth["median_contribution_ltv"],
    linewidth=2,
    linestyle="--",
    label="Медианный Contribution Profit LTV"
)

plt.title(
    "Накопленный Contribution Profit LTV "
    "в зависимости от глубины покупок",
    fontsize=14
)

plt.xlabel("Количество заказов клиента")
plt.ylabel("Накопленный Contribution Profit, ₽")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Клиентская ценность в исследуемой базе формируется прежде всего за счёт высокой частоты повторных покупок. Накопленный Contribution Profit LTV последовательно увеличивается с ростом глубины клиентского цикла и достигает около 81,8 тыс. ₽ к 95-му заказу. При этом после примерно 40–60 заказов темп прироста LTV становится менее выраженным, а после 65 заказов прибыльность отдельного заказа снижается из-за уменьшения размера корзины. Таким образом, высокая частота покупок продолжает обеспечивать рост LTV даже при снижении Contribution Profit на отдельный заказ.
